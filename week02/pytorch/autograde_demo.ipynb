```python
"""
1. requires_grad=True tells PyTorch to track operations on a tensor.

2. PyTorch builds a computational graph automatically.

3. backward() computes gradients using automatic differentiation.

4. The computed gradient is stored in tensor.grad.
"""


import torch
x = torch.tensor(2.0, requires_grad=True)

print(x)
print(x.requires_grad)
y = x ** 2 + 3 * x + 5
print(y)
y.backward()
print(x.grad)
```
