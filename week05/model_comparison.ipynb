```python
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

X = iris.data[:100, [0, 2]]
y = iris.target[:100]

from sklearn.model_selection import train_test_split

# Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1,
    stratify=y
)

X_train_std = (X_train - X_train.mean(axis=0)) / X_train.std(axis=0)
X_test_std = (X_test - X_train.mean(axis=0)) / X_train.std(axis=0)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Create a logistic regression model
lr = LogisticRegression(C=1.0, random_state=1)

lr.fit(X_train_std, y_train)

y_pred = lr.predict(X_test_std)

acc_lr = accuracy_score(y_test, y_pred)

print(acc_lr)

print(lr.coef_)
print(lr.intercept_)

# Predict probabilities for the first three samples in the test set
lr.predict_proba(X_test_std[:3, :])
```

```python
from sklearn.svm import SVC

# Create a Support Vector Machine (SVM) classifier with a linear kernel
svm = SVC(
    kernel='linear',
    C=1.0
)

svm.fit(X_train_std, y_train)

y_pred_svm = svm.predict(X_test_std)

acc_svm = accuracy_score(y_test, y_pred_svm)

print(acc_svm)

# svm也是学习w，b而不是margin
print(svm.coef_)
print(svm.intercept_)

# Predict decision function values for the first three samples in the test set
svm.decision_function(X_test_std[:3, :])
```

```python
svm_small_c = SVC(
    kernel='linear',
    C=0.01
)   

svm_large_c = SVC(
    kernel='linear',
    C=100.0
)
svm_small_c.fit(X_train_std, y_train)
svm_large_c.fit(X_train_std, y_train)

print(
    accuracy_score(y_test, svm_small_c.predict(X_test_std)),
    accuracy_score(y_test, svm_large_c.predict(X_test_std))
)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Create a Decision Tree classifier
tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=1
)

tree.fit(X_train, y_train)

y_pred_tree = tree.predict(X_test)

print(y_pred_tree)

print(accuracy_score(y_test, y_pred_tree))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Create a K-Nearest Neighbors (KNN) classifier
knn = KNeighborsClassifier(
    n_neighbors=5
)

knn.fit(X_train_std, y_train)

y_pred_knn = knn.predict(X_test_std)

print(y_pred_knn)

print(accuracy_score(y_test, y_pred_knn))
```

```python
from sklearn.svm import SVC

# Create a Support Vector Machine (SVM) classifier with a radial basis function (RBF) kernel
rbf_svm = SVC(
    kernel='rbf',
    gamma="scale", 
    C=1.0,
    random_state=1
)   
rbf_svm.fit(X_train_std, y_train)

print("RBF SVM:", accuracy_score(y_test, rbf_svm.predict(X_test_std)))
```

```python
# Week05 Day06 Model Comparison Lab 

lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_std, y_train)

linear_svm = SVC(kernel='linear', C=1.0)
linear_svm.fit(X_train_std, y_train)

rbf_svm = SVC(
    kernel='rbf',
    gamma="scale", 
    C=1.0
)   
rbf_svm.fit(X_train_std, y_train)

tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=1
)
tree.fit(X_train, y_train)

knn = KNeighborsClassifier(
    n_neighbors=5
)
knn.fit(X_train_std, y_train)

# Evaluate the models on the training and test sets
results = {
    "Logistic Regression": (
        lr.score(X_train_std, y_train),
        lr.score(X_test_std, y_test)
    ),
                            

    "Linear SVM": (
        linear_svm.score(X_train_std, y_train),
        linear_svm.score(X_test_std, y_test)
    ),
    "RBF SVM": (
        rbf_svm.score(X_train_std, y_train),
        rbf_svm.score(X_test_std, y_test)
    ),
    "Decision Tree": (
        tree.score(X_train, y_train),
        tree.score(X_test, y_test)
    ),
    "KNN": (
        knn.score(X_train_std, y_train),
        knn.score(X_test_std, y_test)
    )
}

for model, (train_acc, test_acc) in results.items():
    print(
        f"{model:20s} "
        f": Train: {train_acc:.3f} "
        f": Test: {test_acc:.3f} "
    )
```
