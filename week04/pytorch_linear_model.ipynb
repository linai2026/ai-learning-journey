```python
import torch

X = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

print(X.shape)  # torch.Size([4, 1]) 注意不是（4,）而是（4,1）
print(y.shape)  # torch.Size([4, 1])

w = torch.tensor([[0.0]], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

print(w.shape)  # torch.Size([1, 1])
print(b.shape)  # torch.Size([1])

# Forward Pass
y_pred = X @ w + b
print(y_pred)
print(y_pred.shape)  # torch.Size([4, 1])

# Compute Loss
error = y_pred - y
print(error)
loss = (error ** 2).mean()
print(loss)

```

```python
import torch

X = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

w = torch.tensor([[0.0]], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

learning_rate = 0.01

for epoch in range(500):

    # 每个epoch都要进行前向传播，计算损失，反向传播，更新参数，清零梯度

    # Forward Pass 
    y_pred = X @ w + b

    # Compute Loss
    loss = ((y_pred - y) ** 2).mean()


    # Backward Pass 只计算梯度不更新参数
    loss.backward()

    # Update Parameters
    with torch.no_grad(): #告诉 pytorch 下面是Update Parameters 不需要建立gradient tracking
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad
    

    # Zero Gradients pytorch默认会累积梯度，所以每次更新完参数后需要清零梯度
    w.grad.zero_()
    b.grad.zero_()

    if epoch % 50 == 0:
        #loss.item()把tensor转换为python的float类型，方便打印记录
        print(epoch, loss.item())
```
