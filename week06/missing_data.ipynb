```python
import numpy as np
import pandas as pd

# Create a DataFrame with missing values
df = pd.DataFrame({
    "height": [170, 180, np.nan, 175, 168],
    "weight": [65, np.nan, 55, 73, 62],
    "age": [30, 41, 25, np.nan, 28]
})

# Check for missing values
print(df.isnull())
print(df.isnull().sum())

# Drop rows with missing values
df.dropna()
# Drop columns with missing values
df.dropna(axis=1)

# 缺失值不参与 mean
df.mean()
```

```python
from sklearn.impute import SimpleImputer

# Create an imputer object with a mean filling strategy
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")

# Fit the imputer on the DataFrame，学习 mean of each feature (column) in the DataFrame
imputer.fit(df)
imputer.statistics_  # 查看每一列的均值

# Transform the DataFrame to fill in missing values
X_imputed = imputer.transform(df)

type(X_imputed) # numpy.ndarray 不是原来的 pandas DataFrame

df_imputed = pd.DataFrame(X_imputed, columns = df.columns)
type(df_imputed) #pandas.DataFrame

```
