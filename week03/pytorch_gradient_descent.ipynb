```python
import torch

x = torch.tensor(0.0, requires_grad=True)

for i in range(10):
    # Forward pass  
    y = (x - 3) ** 2

    # Backward pass
    y.backward()
    
    print(x,y,x.grad)

    # Update parameters using gradient descent
    with torch.no_grad():
        x -= 0.1 * x.grad

    # Zero the gradients after updating
    x.grad.zero_()

```
